```python
import numpy as np

# Arrhenius rate constants
Ea = 75e3   # J/mol
A = 1e13    # 1/s
R = 8.314
T = np.array([300, 400, 500, 600])
k = A * np.exp(-Ea / (R * T))
for Ti, ki in zip(T, k):
    print(f'{Ti} K: k = {ki:.3e} 1/s')
```

The rate constant goes up by more than six orders of magnitude between 300 K and 600 K.
